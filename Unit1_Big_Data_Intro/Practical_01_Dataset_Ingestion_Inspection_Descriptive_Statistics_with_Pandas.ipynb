# Practical 01: Dataset Ingestion, Inspection & Descriptive Statistics with Pandas

**UNIT 1: Introduction to Big Data and Large Dataset Handling**

**Problem Statement:** Write a Python program using Pandas to load a structured dataset (CSV/JSON), display its first 5 rows, last 5 rows, dimensions (shape), column names, and data types with info(). Generate comprehensive summary statistics of all numeric and categorical attributes using describe(include='all').

```python
import io
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None, "display.width", 200)

rs = np.random.default_rng(11)
n = 250
base = pd.DataFrame({
    "Order ID": np.arange(1001, 1001 + n),
    "Customer Name": [f"Cust-{i:03d}" for i in rs.integers(1, 80, n)],
    "City": rs.choice(["Kochi", "Chennai", "Mumbai", "Bengaluru", "Hyderabad", "Pune"], n),
    "Product Category": rs.choice(["Electronics", "Furniture", "Groceries", "Sports", "Books"], n),
    "Quantity": rs.integers(1, 8, n),
    "Unit Price": np.round(rs.uniform(200, 6000, n), 2),
    "Discount %": rs.choice([0, 10, 15, 20, 30], n),
    "Order Date": pd.date_range("2025-06-01", periods=n, freq="D").strftime("%Y-%m-%d"),
})

csv_buf = io.StringIO()
base.to_csv(csv_buf, index=False)
csv_buf.seek(0)
json_buf = io.StringIO(base.to_json(orient="records"))

data = pd.read_csv(csv_buf)
print("Dataset loaded successfully!")

print(data.head())

print(data.tail())

print("Shape (rows, columns):", data.shape)
print("Column names:", data.columns.tolist())

data.info()

print(data.describe(include='all'))

df_from_json = pd.read_json(json_buf)
print("JSON load – shape:", df_from_json.shape)
print(df_from_json.head())
```
