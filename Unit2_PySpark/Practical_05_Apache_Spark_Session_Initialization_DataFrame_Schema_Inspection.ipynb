# Practical 05: Apache Spark Session Initialization & DataFrame Schema Inspection

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a Python program to install and initialize an Apache Spark session in PySpark using SparkSession.builder. Create a structured PySpark DataFrame from Python dictionaries, display its schema tree hierarchy using printSchema(), and print its rows using .show().

```python
from pyspark.sql import SparkSession, Row

spark = (SparkSession.builder
         .appName("DS505-P05")
         .master("local[*]")
         .config("spark.ui.showConsoleProgress", "false")
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

print("Spark version:", spark.version)

data = [
    {"Name": "Arjun", "Department": "IT", "City": "Pune", "Salary": 74000, "Experience": 6},
    {"Name": "Sneha", "Department": "HR", "City": "Delhi", "Salary": 56000, "Experience": 3},
    {"Name": "Vivek", "Department": "Finance", "City": "Pune", "Salary": 67000, "Experience": 7},
    {"Name": "Kriti", "Department": "IT", "City": "Jaipur", "Salary": 90000, "Experience": 9},
    {"Name": "Manav", "Department": "Sales", "City": "Delhi", "Salary": 49000, "Experience": 2},
]
emp_df = spark.createDataFrame([Row(**d) for d in data])
emp_df.printSchema()
emp_df.show()
spark.stop()
```
