# Practical 06: PySpark DataFrame Transformations: Selection, Filtering & Columns

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a PySpark program to execute DataFrame transformations: select specific columns using .select(), filter rows using multi-condition boolean logic (e.g., Salary > 60000 and City == 'Surat') using .filter(), create a computed column using .withColumn(), and rename a column using .withColumnRenamed().

```python
from pyspark.sql import SparkSession, Row
from pyspark.sql.functions import col

spark = (SparkSession.builder.appName("DS505-P06").master("local[*]")
         .config("spark.ui.showConsoleProgress", "false").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

data = [
    {"Name": "Arjun", "Department": "IT", "City": "Pune", "Salary": 74000},
    {"Name": "Sneha", "Department": "HR", "City": "Delhi", "Salary": 56000},
    {"Name": "Vivek", "Department": "Finance", "City": "Pune", "Salary": 67000},
    {"Name": "Kriti", "Department": "IT", "City": "Jaipur", "Salary": 90000},
    {"Name": "Manav", "Department": "Sales", "City": "Delhi", "Salary": 49000},
    {"Name": "Aditi", "Department": "IT", "City": "Pune", "Salary": 95000},
]
emp_df = spark.createDataFrame([Row(**d) for d in data])
emp_df.show()

emp_df.select("Name", "City", "Salary").show()

filtered = emp_df.filter((col("Salary") > 62000) & (col("City") == "Pune"))
filtered.show()

with_bonus = emp_df.withColumn("Annual_Bonus", col("Salary") * 0.12)
with_bonus.show()

renamed = with_bonus.withColumnRenamed("Name", "Employee_Name")
renamed.show()
renamed.printSchema()
spark.stop()
```
