# Practical 07: Grouped Aggregations & Action Execution in PySpark

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a PySpark program to perform grouped aggregations and action executions. Group data by Department using .groupBy(), calculate aggregate metrics (count, average salary, maximum salary, total salary) using pyspark.sql.functions, sort records using .orderBy(), and execute actions (.count(), .show()).

```python
from pyspark.sql import SparkSession, Row
from pyspark.sql import functions as F

spark = (SparkSession.builder.appName("DS505-P07").master("local[*]")
         .config("spark.ui.showConsoleProgress", "false").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

data = [
    {"Name": "Arjun", "Department": "IT", "Salary": 74000},
    {"Name": "Sneha", "Department": "HR", "Salary": 56000},
    {"Name": "Vivek", "Department": "Finance", "Salary": 67000},
    {"Name": "Kriti", "Department": "IT", "Salary": 90000},
    {"Name": "Manav", "Department": "Sales", "Salary": 49000},
    {"Name": "Aditi", "Department": "IT", "Salary": 95000},
    {"Name": "Rahul", "Department": "HR", "Salary": 60000},
    {"Name": "Pallavi", "Department": "Finance", "Salary": 79000},
    {"Name": "Sahil", "Department": "Sales", "Salary": 51000},
]
emp_df = spark.createDataFrame([Row(**d) for d in data])
emp_df.show()

summary = (emp_df.groupBy("Department")
             .agg(F.count("*").alias("Employee_Count"),
                  F.round(F.avg("Salary"), 2).alias("Avg_Salary"),
                  F.max("Salary").alias("Max_Salary"),
                  F.sum("Salary").alias("Total_Salary"))
             .orderBy(F.col("Avg_Salary").desc()))

summary.show()

print("Total employees      (emp_df.count())     :", emp_df.count())
print("Number of departments (summary.count()):", summary.count())

spark.stop()
```
