# Practical 12: Classification with Decision Trees & K-Nearest Neighbors (KNN)

**UNIT 3: Preparing Big Data for Machine Learning**

**Problem Statement:** Write a Python program to build and compare two classification models: DecisionTreeClassifier and KNeighborsClassifier (KNN) using Scikit-Learn. Train both models on a classification dataset, evaluate test predictions, and compare their performance using Confusion Matrix and Accuracy Score.

```python
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import confusion_matrix, accuracy_score

X, y = make_classification(n_samples=1500, n_features=14, n_informative=7, n_redundant=2,
                           n_classes=2, class_sep=0.9, flip_y=0.05, random_state=1)
print("Dataset shape:", X.shape)
print("Class counts :", dict(pd.Series(y).value_counts()))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1, stratify=y)
print("Train:", X_train.shape, "| Test:", X_test.shape)

candidates = {
    "Decision Tree": DecisionTreeClassifier(max_depth=6, random_state=1),
    "KNN (k=3)": make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=3)),
}

scores = {}
for name, model in candidates.items():
    model.fit(X_train, y_train)
    y_hat = model.predict(X_test)
    acc = accuracy_score(y_test, y_hat)
    scores[name] = acc

    print(f"===== {name} =====")
    print("Confusion matrix (rows = actual, columns = predicted):")
    print(pd.DataFrame(confusion_matrix(y_test, y_hat),
                         index=["Actual 0", "Actual 1"], columns=["Pred 0", "Pred 1"]))
    print(f"Accuracy: {acc:.4f}\n")

comparison = (pd.Series(scores, name="Accuracy").sort_values(ascending=False).to_frame())
print(comparison)
print("Best model:", comparison.index[0])
```
