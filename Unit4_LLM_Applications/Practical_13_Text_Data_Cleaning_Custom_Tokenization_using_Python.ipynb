**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to clean raw, noisy text data: convert characters to lowercase using lower(), remove special punctuation symbols using replace(), eliminate redundant whitespace using strip(), and tokenize sentences into individual word tokens using split().

```python
import string

noisy_samples = [
    "   Cloud storage is CHEAP,,, but data-cleaning is still   HARD!!!   ",
    "Deep-Learning models need LOTS of data (and lots of GPUs)...  ",
    "   Visit https://example.com/docs ??? Price: $99.99 only!   ",
]
for t in noisy_samples:
    print(repr(t))

def normalise(text: str) -> str:
    text = text.lower()
    for p in string.punctuation:
        text = text.replace(p, "")
    text = " ".join(text.split())
    return text.strip()

def split_tokens(text: str) -> list:
    return text.split()

for i, raw in enumerate(noisy_samples, 1):
    cleaned = normalise(raw)
    tokens = split_tokens(cleaned)
    print(f"--- Sentence {i} ---")
    print("Raw     :", repr(raw))
    print("Cleaned :", repr(cleaned))
    print("Tokens  :", tokens)
    print("Count   :", len(tokens), "\n")
```
