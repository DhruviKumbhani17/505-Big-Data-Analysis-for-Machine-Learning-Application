# Practical 14: Vocabulary Frequency Profiling & Stopword Filtering

**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to profile vocabulary in a text corpus: clean sentences, filter out common English stopwords (e.g., 'the', 'is', 'in', 'and'), calculate word occurrence frequencies using collections.Counter, and display the top 5 most frequent keywords.

```python
import string
from collections import Counter

corpus = [
    "Data engineers build pipelines that move data from sensors into the warehouse.",
    "The warehouse stores the data in a columnar format for fast queries.",
    "A clean dataset is the first step of every machine learning project.",
    "Models are tested on new data to measure how well they generalise.",
    "Large language models read text data and predict the next word.",
]

STOP_WORDS = {"the", "a", "an", "is", "are", "in", "of", "to", "for", "into", "from", "on", "at", "by", "that", "as", "new", "first", "every"}

def clean(sentence: str) -> str:
    sentence = sentence.lower()
    for p in string.punctuation:
        sentence = sentence.replace(p, "")
    return " ".join(sentence.split())

tokens = []
for s in corpus:
    tokens.extend(clean(s).split())

kept = [w for w in tokens if w not in STOP_WORDS]

print(f"Total tokens (before filtering): {len(tokens)}")
print(f"Total tokens (after  filtering): {len(kept)}")
print(f"Vocabulary size (unique words) : {len(set(kept))}")

counts = Counter(kept)
print(counts)

for rank, (word, count) in enumerate(counts.most_common(6), 1):
    print(f"{rank}. {word:<10} {count}  {'#' * count}")
```
